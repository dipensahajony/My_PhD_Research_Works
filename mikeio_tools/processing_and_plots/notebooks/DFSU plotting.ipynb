```python
import mikeio
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.tri as tri
from mpl_toolkits.axes_grid1 import make_axes_locatable

# File paths
wave_height_file = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\Small_scale_Wave_Model_Cyclone_Harvey.sw - Result Files\small_scale_wave_result_Stat_wv_ht.dfsu"
wave_power_file = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\Small_scale_Wave_Model_Cyclone_Harvey.sw - Result Files\small_scale_wave_result_Stat_wv_power.dfsu"
rad_stress_file = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\Small_scale_Wave_Model_Cyclone_Harvey.sw - Result Files\small_scale_wave_result_Stat_rad_stress.dfsu"

# Item names
item_name1 = "Statistical maximum : Sign. Wave Height"
item_name2 = "Statistical maximum : Wave power, p"
item_name3 = "Statistical maximum : Rad. Stress, Sxy"

# Read DataArrays
da_wvht = mikeio.read(wave_height_file, items=[item_name1])[0][0]
da_wvpower = mikeio.read(wave_power_file, items=[item_name2])[0][0]
da_radstress = mikeio.read(rad_stress_file, items=[item_name3])[0][0]

# Mesh info
mesh = da_wvht.geometry
nodes = mesh.node_coordinates
elements_raw = mesh.element_table

# Keep only triangles
tri_elements = np.array([elem for elem in elements_raw if len(elem) == 3], dtype=np.int32)
triang = tri.Triangulation(nodes[:, 0], nodes[:, 1], tri_elements)

# Trim values to triangle count
values1 = da_wvht.values[:len(tri_elements)]
values2 = da_wvpower.values[:len(tri_elements)]
values3 = da_radstress.values[:len(tri_elements)]

# Setup figure
fig, axs = plt.subplots(1, 3, figsize=(21, 7))

# Plot helper function
def plot_panel(ax, triang, values, title, cbar_label):
    pcm = ax.tripcolor(triang, values, shading="flat", cmap="viridis", edgecolors='none')
    ax.set_title(title, fontsize=13)
    ax.set_xlabel("X [m]")
    ax.set_ylabel("Y [m]")
    ax.set_aspect('equal')
    # Custom colorbar with matching height
    divider = make_axes_locatable(ax)
    cax = divider.append_axes("right", size="5%", pad=0.05)
    cb = fig.colorbar(pcm, cax=cax)
    cb.set_label(cbar_label, fontsize=11)

# Plot all three
plot_panel(axs[0], triang, values1, "Existing condition - Wave Height", "Wave Height (m)")
plot_panel(axs[1], triang, values2, "Existing condition - Wave Power", "Wave Power (W/m)")
plot_panel(axs[2], triang, values3, "Existing condition - Radiation Stress", "Radiation Stress Sxy (N/m²)")

plt.tight_layout()
plt.show()
```

```python
import mikeio
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1 import make_axes_locatable

# File paths
wave_height_file = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\east_side_barrier_reef.sw - Result Files\east_side_reef_Stat_wv_ht.dfsu"
wave_power_file  = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\east_side_barrier_reef.sw - Result Files\east_side_reef_Stat_wv_power.dfsu"
rad_stress_file  = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\east_side_barrier_reef.sw - Result Files\east_side_reef_Stat_rad_stress.dfsu"

# Variables to plot
item1 = "Statistical maximum : Sign. Wave Height"
item2 = "Statistical maximum : Wave power, p"
item3 = "Statistical maximum : Rad. Stress, Sxy"

# Read datasets
ds_wvht      = mikeio.read(wave_height_file, items=[item1])
ds_wvpower   = mikeio.read(wave_power_file,  items=[item2])
ds_radstress = mikeio.read(rad_stress_file,  items=[item3])

# Increase all text sizes
plt.rcParams.update({
    'font.size': 16,
    'axes.titlesize': 20,
    'axes.labelsize': 18,
    'xtick.labelsize': 16,
    'ytick.labelsize': 16
})

# Create subplots
fig, axs = plt.subplots(1, 3, figsize=(21, 9))

# Plot each without the automatic colorbar
_ = ds_wvht[0].plot(ax=axs[0], add_colorbar=False, title="Reef Condition – Wave Height")
_ = ds_wvpower[0].plot(ax=axs[1], add_colorbar=False, title="Reef Condition – Wave Power")
_ = ds_radstress[0].plot(ax=axs[2], add_colorbar=False, title="Reef Condition – Radiation Stress")

# Manually add a colorbar matching each plot's height
for ax, im, label in zip(
    axs,
    (axs[0].collections[0], axs[1].collections[0], axs[2].collections[0]),
    ("Significant Wave Height (m)", "Wave Power (kW/m)", "Radiation Stress (N/m²)")
):
    divider = make_axes_locatable(ax)
    cax = divider.append_axes("right", size="5%", pad=0.02)
    cbar = plt.colorbar(im, cax=cax)
    cbar.set_label(label, fontsize=16)
    cbar.ax.tick_params(labelsize=14)

# Increase spacing between x‐axis tick labels (show fewer ticks)
for ax in axs:
    ticks = ax.get_xticks()
    ax.set_xticks(ticks[::3])       # show one label every 3 ticks; adjust slice as needed
    ax.xaxis.set_tick_params(pad=10) # vertical gap between axis and labels

plt.tight_layout()
plt.show()
```
