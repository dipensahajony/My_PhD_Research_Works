```python
import mikeio
import matplotlib.pyplot as plt
import numpy as np

# File paths
wave_ht_file = r"D:\Phd Research\Wave Model\small scale port lavaca\Stat_dfsu\max_sig_wave_ht_line.dfs1"
bathy_file = r"D:\Phd Research\Wave Model\small scale port lavaca\Stat_dfsu\Cross-shore_bathymetry.dfs1"

# Read DFS1 files
wave_dfs = mikeio.Dfs1(wave_ht_file)
bathy_dfs = mikeio.Dfs1(bathy_file)

# Read data
wave_data = wave_dfs.read()
bathy_data = bathy_dfs.read()

# Print available items to confirm correct variable selection
# print("Wave Height Variables:", wave_dfs.items)
# print("Bathymetry Variables:", bathy_dfs.items)

# Select the first variable in each file (modify if needed)
wave_item_index = 1
bathy_item_index = 0

# Convert to NumPy arrays and flatten
wave_values = wave_data[wave_item_index].to_numpy().flatten()
bathy_values = bathy_data[bathy_item_index].to_numpy().flatten()

# Define X-axis as 0 to (N-1), assuming both files have the same number of points
num_points = len(wave_values)
x_axis = np.arange(num_points)

# Plot wave height and bathymetry
fig, ax1 = plt.subplots(figsize=(8, 5))

# Plot wave height (left Y-axis)
ax1.plot(x_axis, wave_values, marker="o", linestyle="-", color="b", linewidth=2, label="Max Sig. Wave Height")
ax1.set_xlabel("Point Index along Line (0 to 9)")
ax1.set_ylabel(f"{wave_dfs.items[wave_item_index].name} (m)", color="b")
ax1.tick_params(axis="y", labelcolor="b")

# Plot bathymetry (right Y-axis)
ax2 = ax1.twinx()
ax2.plot(x_axis, bathy_values, marker="s", linestyle="--", color="brown", linewidth=2, label="Bathymetry")
ax2.set_ylabel(f"{bathy_dfs.items[bathy_item_index].name} Elevation(m)", color="brown")
ax2.tick_params(axis="y", labelcolor="brown")

# Grid and title
plt.title("Max Significant Wave Height & Bathymetry Profile")
ax1.grid(True)

# Show legend
fig.legend(loc="center")

# Show plot
plt.show()
```

```python
import mikeio
import matplotlib.pyplot as plt
import numpy as np

# File paths
wave_ht_file = r"D:\Phd Research\Wave Model\small scale port lavaca\Stat_dfsu\max_sig_wave_ht_line.dfs1"
wave_ht_file_altered = r"D:\Phd Research\Wave Model\small scale port lavaca\Stat_dfsu\max_sig_wave_ht_line_altered.dfs1"
bathy_file = r"D:\Phd Research\Wave Model\small scale port lavaca\Stat_dfsu\Cross-shore_bathymetry.dfs1"

# Read DFS1 files
wave_dfs = mikeio.open(wave_ht_file)
wave_dfs_altered = mikeio.open(wave_ht_file_altered)
bathy_dfs = mikeio.open(bathy_file)

# Select the first variable in each file
wave_item_index = 1  # Adjust index if needed
bathy_item_index = 0

# Read data
wave_data = wave_dfs.read(items=[wave_item_index])
wave_data_altered = wave_dfs_altered.read(items=[wave_item_index])
bathy_data = bathy_dfs.read(items=[bathy_item_index])

# Convert to NumPy arrays
wave_values = wave_data[0].to_numpy().flatten()
wave_values_altered = wave_data_altered[0].to_numpy().flatten()
bathy_values = bathy_data[0].to_numpy().flatten()

# Define X-axis (assuming same number of points in all datasets)
num_points = len(wave_values)
x_axis = np.arange(num_points)

# Create figure and axes
fig, ax1 = plt.subplots(figsize=(8, 5))

# Plot original wave height (left Y-axis)
ax1.plot(x_axis, wave_values, marker="o", linestyle="-", color="b", linewidth=2, label="Max Sig. Wave Height (Original)")
ax1.plot(x_axis, wave_values_altered, marker="^", linestyle="--", color="r", linewidth=2, label="Max Sig. Wave Height (Altered)")
ax1.set_xlabel("Point Index along Line")
ax1.set_ylabel(f"{wave_dfs.items[wave_item_index].name} (m)", color="b")
ax1.tick_params(axis="y", labelcolor="b")

# Plot bathymetry (right Y-axis)
ax2 = ax1.twinx()
ax2.plot(x_axis, bathy_values, marker="s", linestyle="--", color="brown", linewidth=2, label="Bathymetry")
ax2.set_ylabel(f"{bathy_dfs.items[bathy_item_index].name} Elevation (m)", color="brown")
ax2.tick_params(axis="y", labelcolor="brown")

# Add legends separately for each axis
ax1.legend(loc="center")
ax2.legend(loc="center")

# Grid and title
plt.title("Max Significant Wave Height & Bathymetry Profile")
ax1.grid(True)

# Show plot
plt.show()
```
