```python
import mikeio
import numpy as np

# -------- USER INPUT --------
file1 = r"D:\Phd Research\Prototype HD Model\Simulation\Without Reef Scenarios\Climate change condition\100yr_compound_flood_base.m21fm - Result Files\100yr_compound_flood_base_stat.dfsu"
file2 = r"D:\Phd Research\Prototype HD Model\Simulation\Without Reef Scenarios\Climate change condition\100yr_compound_flood_base_Q_P20.m21fm - Result Files\100yr_compound_flood_base_Q_P_20_stat.dfsu"
# ----------------------------

# 1. Open both files
ds1 = mikeio.open(file1)
ds2 = mikeio.open(file2)

print("File 1 items:", ds1.items)
print("File 2 items:", ds2.items)

# 2. Compare geometry
geom_equal = ds1.geometry == ds2.geometry
print(f"Same mesh geometry? {geom_equal}")

# 3. Load data
data1 = ds1.read()
data2 = ds2.read()

# Ensure they have same number of items & timesteps
if len(data1.items) != len(data2.items):
    print("Different number of items!")
if len(data1.time) != len(data2.time):
    print("Different number of timesteps!")

# 4. Compare item by item
for i, item in enumerate(data1.items):
    if item.name not in [it.name for it in data2.items]:
        print(f"Item {item.name} not found in File2")
        continue
    arr1 = data1[item.name].to_numpy()
    arr2 = data2[item.name].to_numpy()

    if arr1.shape != arr2.shape:
        print(f"Item {item.name}: Different shapes {arr1.shape} vs {arr2.shape}")
    else:
        diff = np.nanmax(np.abs(arr1 - arr2))
        if diff == 0:
            print(f"Item {item.name}: Identical")
        else:
            print(f"Item {item.name}: Max abs difference = {diff:.6f}")

print("Comparison complete.")
```
