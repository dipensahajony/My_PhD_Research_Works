```python
import os
import pandas as pd

# Folder containing the .xyz files
folder_path = r"D:\Phd Research\Prototype HD Model\XYZ file\Engineered reef\Port_Lavaca\Port_lavaca_0m_height"

# Loop through each file in the folder
for filename in os.listdir(folder_path):
    if filename.endswith('.xyz'):
        file_path = os.path.join(folder_path, filename)
        
        # Read the xyz file as a DataFrame, assuming a whitespace delimiter
        df = pd.read_csv(file_path, delim_whitespace=True, header=None)
        
        # Change the third column (index 2) to 2
        df[2] = 0
        
        # Save the modified file
        df.to_csv(file_path, sep=' ', header=False, index=False)

print("All files have been updated.")
```
