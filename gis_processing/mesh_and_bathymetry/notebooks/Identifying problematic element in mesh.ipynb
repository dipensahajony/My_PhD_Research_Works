```python
import mikeio
import matplotlib.pyplot as plt

# Load your mesh file
mesh_file = r"D:\Phd Research\Prototype HD Model\Mesh File\prototype_domain_utm_corrected_mesh_for_100yr.mesh"  # replace with actual path
mesh = mikeio.Mesh(mesh_file)

# Target element index (from error)
target_elem = 32438

# Get coordinates of the element center
x_center = mesh.element_coordinates[target_elem][0]
y_center = mesh.element_coordinates[target_elem][1]

print(f"Element {target_elem} center: X = {x_center}, Y = {y_center}")
```

```python
# Plot entire mesh
plt.figure(figsize=(10, 8))
plt.triplot(mesh.node_coordinates[:, 0], mesh.node_coordinates[:, 1], mesh.element_table, linewidth=0.5, color='lightgrey')

# Highlight the problematic element
element_nodes = mesh.element_table[target_elem]
node_coords = mesh.node_coordinates[element_nodes]

# Close the polygon by repeating the first point
x_coords = list(node_coords[:, 0]) + [node_coords[0, 0]]
y_coords = list(node_coords[:, 1]) + [node_coords[0, 1]]

plt.fill(x_coords, y_coords, color='red', alpha=0.5, label=f'Element {target_elem}')
plt.plot(x_center, y_center, 'ko', label='Element Center')

plt.xlabel("X [m]")
plt.ylabel("Y [m]")
plt.title("Problematic Element in Mesh")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
