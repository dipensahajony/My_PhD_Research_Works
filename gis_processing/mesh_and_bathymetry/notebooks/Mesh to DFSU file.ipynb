```python
import mikeio

# Load the mesh
mesh = mikeio.Mesh(r"D:\Phd Research\Wave Model\small scale port lavaca\Mesh file\Altered_bathymetry_original.mesh")

# Extract node coordinates and element table from the mesh
node_coords = mesh.node_coordinates  # Node coordinates (x, y)
elements = mesh.element_table        # Element connectivity

# If bathymetry exists in the mesh, extract it
bathymetry = mesh.zn if hasattr(mesh, 'zn') else None

# Create a new DFSU object
dfsu = mikeio.dfsu.Dfsu.from_array(
    filename=r"D:\Phd Research\Wave Model\small scale port lavaca\Mesh file\output_dfsu_file.dfsu",
    geometry=mesh.geometry,  # Use mesh geometry directly
    item_names=["Bathymetry"],  # Add item names (you can add more if needed)
    data={ "Bathymetry": bathymetry }  # Add bathymetry data
)

print("DFSU file created successfully!")


```
