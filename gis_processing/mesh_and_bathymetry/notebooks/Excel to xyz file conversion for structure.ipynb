```python
import pandas as pd
import numpy as np
file_path = r"D:\Phd Research\Prototype HD Model\Engineered reef segment\Port_Lavaca_reef_segment.xlsx"
sheet_name = "Corpus Christi"
data = pd.read_excel(file_path, sheet_name = sheet_name)
```

```python
for orig_fid in data['ORIG_FID'].unique():
    filtered_data = data[data['ORIG_FID'] == orig_fid]
    # Extract POINT_X, POINT_Y, and elv columns
    xyz_data = filtered_data[['POINT_X', 'POINT_Y', 'elv']]
    # Define the output file path
    output_file_path = f"D:\Phd Research\Prototype HD Model\XYZ file\Engineered reef\Corpus_Christi\Corpus_Christi{orig_fid + 1}.xyz"
    xyz_data.to_csv(output_file_path, header=False, index=False, sep=' ')
    
    print(f"Saved: {output_file_path}")
    
```
