```python
et_ids = data["ET_ID"].unique()
xs = [f"CS-{i+1}" for i in range(et_ids.shape[0])]
xdf = pd.DataFrame({"ET_ID": et_ids, "XS": xs})
data = pd.merge(left=data, right=xdf, on="ET_ID", how="inner")

river_data_list = []
for river in data["RvName"].unique():
    river_data = data[data["RvName"] == river].copy()
    ids = river_data["ET_ID"].unique()
    chainage = list(range(0, len(ids) * 200, 200))
    river_data["Chainage"] = river_data['ET_ID'].map(dict(zip(ids, chainage)))

    river_data["Distance"] = 0
    for i in ids:
        idx = river_data[river_data['ET_ID'] == i].index
        river_data.loc[idx, "Distance"] = list(range(0, idx.shape[0]))

    river_data_list.append(river_data)

formatted_data = pd.concat(river_data_list, ignore_index=True)
formatted_data.to_excel("formatted_data.xlsx", index=False)
```
