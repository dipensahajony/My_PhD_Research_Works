```python
import pandas as pd

# List of file paths
file_paths = [
    r"D:\Phd Research\Prototype HD Model\XYZ file\Barrier Island\island_1.xyz",
    r"D:\Phd Research\Prototype HD Model\XYZ file\Barrier Island\island_2.xyz",
    r"D:\Phd Research\Prototype HD Model\XYZ file\Barrier Island\island_3.xyz",
    r"D:\Phd Research\Prototype HD Model\XYZ file\Barrier Island\island_4.xyz",
    r"D:\Phd Research\Prototype HD Model\XYZ file\Barrier Island\island_5.xyz",
    r"D:\Phd Research\Prototype HD Model\XYZ file\Barrier Island\island_6.xyz"
]

# Read and combine all files
df_all = pd.concat(
    [pd.read_csv(fp, delim_whitespace=True, header=None, names=["X", "Y", "Z"]) for fp in file_paths],
    ignore_index=True
)

# Find min and max of Z
z_min = df_all["Z"].min()
z_max = df_all["Z"].max()

print(f"Minimum Z value: {z_min}")
print(f"Maximum Z value: {z_max}")
```
