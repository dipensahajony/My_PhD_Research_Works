```python
import mikeio
import geopandas as gpd
from shapely.geometry import Polygon

# Load the MIKE mesh file
mesh_path = r"D:\Phd Research\Wave Model\small scale port lavaca\Mesh file\updated_bathymetry_west_side.mesh"
mesh = mikeio.Mesh(mesh_path)
elements = mesh.element_table
nodes = mesh.node_coordinates

# Create polygons for each triangle element
element_polygons = []
for tri in elements:
    coords = [nodes[i] for i in tri]
    poly = Polygon(coords)
    element_polygons.append(poly)

# Create GeoDataFrame for all elements
mesh_gdf = gpd.GeoDataFrame({'geometry': element_polygons}, crs="EPSG:32614")  # Adjust CRS if needed

# Save all elements to shapefile
output_path = r"D:\Phd Research\Prototype HD Model\Mesh File\all_mesh_small_scale.shp"
mesh_gdf.to_file(output_path)

print(f"✅ All mesh elements saved to: {output_path}")
```
