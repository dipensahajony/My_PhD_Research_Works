```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point

# Load Excel file
file_path = r"D:\Phd Research\Prototype HD Model\Simulation\Without Reef Scenarios\Climate change condition\100-yr return period storm CC.m21fm - Result Files\xy_mx_total_water_depth.xlsx"
df = pd.read_excel(file_path)

# Inspect column names to ensure correct mapping
df.columns

# Assume columns: X, Y, Value (rename if different)
x_col = 'X'  # replace with exact column name
y_col = 'Y'  # replace with exact column name
value_col = 'Water_depth'  # replace with correct column name

# Create geometry from coordinates
geometry = [Point(xy) for xy in zip(df[x_col], df[y_col])]

# Create GeoDataFrame with correct CRS (EPSG:32614)
gdf = gpd.GeoDataFrame(df, geometry=geometry, crs="EPSG:32614")

# Output shapefile
output_path = r"D:\Phd Research\total_water_depth_points.shp"
gdf.to_file(output_path)

print("Shapefile saved to:", output_path)
```
