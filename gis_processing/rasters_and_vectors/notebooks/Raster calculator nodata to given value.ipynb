```python
import rasterio
import numpy as np

# --- Input / Output paths ---
raster_in  = r"C:\Users\sahad2\Research_data\Raster\diff_com100_Surge_modified.tif"
raster_out = r"D:\Phd Research\diff_com100_Surge_modified_NoNaN.tif"

# --- Open the raster ---
with rasterio.open(raster_in) as src:
    data = src.read(1)  # read first band
    profile = src.profile
    nodata = src.nodata

# --- Replace NoData with 0 ---
if nodata is not None:
    data[data == nodata] = 0.001
else:
    # Also handle NaN if nodata was not defined
    data[np.isnan(data)] = 0.001

# --- Update profile (optional: remove nodata flag) ---
profile.update(nodata=0)

# --- Write output raster ---
with rasterio.open(raster_out, "w", **profile) as dst:
    dst.write(data, 1)

print("✅ NoData values replaced by 0 and saved to:", raster_out)
```
