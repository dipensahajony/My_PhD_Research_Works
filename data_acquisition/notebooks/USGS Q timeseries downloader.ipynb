```python
import hydrofunctions as hf

def download_usgs_daily_discharge(site, start_date, end_date):
    """
    Download USGS daily discharge data for a specific site and time period.
    
    Parameters:
        site (str): USGS site number.
        start_date (str): Start date in 'YYYY-MM-DD' format.
        end_date (str): End date in 'YYYY-MM-DD' format.
        
    Returns:
        pandas.DataFrame: DataFrame containing daily discharge data.
    """
    data = hf.NWIS(site, "dv", start_date, end_date)
    discharge = data.df()
    return discharge

def create_mike_zero_timeseries(df, filename):
    """
    Create a MIKE Zero time series file from a DataFrame containing discharge data.
    
    Parameters:
        df (pandas.DataFrame): DataFrame containing discharge data.
        filename (str): Name of the output MIKE Zero time series file.
    """
    with open(filename, 'w') as f:
        f.write('Dfs0.TimeSeries.1\n')
        f.write('DataType  Float\n')
        f.write('Dtype      TC_Dynamic_PrecFlooding\n')
        f.write('NumberOfItems 2\n')
        f.write('StartTime 2024-01-01 00:00:00\n')  # Modify start time as needed
        f.write(f'ItemCount {len(df)}\n')
        f.write('Time\tWater Level (m3/s)\n')
        
        for index, row in df.iterrows():
            f.write(f'{index.strftime("%Y-%m-%d")} {row["USGS:08212400:00060:00001"]}\n')


# Example usage:
site = '08212400'  # Replace 'site_number_here' with the USGS site number
start_date = '2024-01-01'
end_date = '2024-01-31'
output_filename = 'discharge.dfs0'

# Download discharge data from USGS
discharge_data = download_usgs_daily_discharge(site, start_date, end_date)

# Create MIKE Zero time series file
create_mike_zero_timeseries(discharge_data, output_filename)
```

```python
import hydrofunctions as hf
import pandas as pd

def download_usgs_daily_discharge(site, start_date, end_date):
    """
    Download USGS daily discharge data for a specific site and time period.
    
    Parameters:
        site (str): USGS site number.
        start_date (str): Start date in 'YYYY-MM-DD' format.
        end_date (str): End date in 'YYYY-MM-DD' format.
        
    Returns:
        pandas.DataFrame: DataFrame containing daily discharge data.
    """
    data = hf.NWIS(site, "dv", start_date, end_date)
    discharge = data.df()
    return discharge

# Example usage:
site = '08212400'  # Replace 'site_number_here' with the USGS site number
start_date = '2023-01-01'
end_date = '2023-12-31'

# Download discharge data from USGS
discharge_data = download_usgs_daily_discharge(site, start_date, end_date)

# Selecting the column containing discharge data
discharge_column = discharge_data.columns[discharge_data.columns.str.contains('00060:00003')].tolist()

# Check if discharge_column is not empty
if discharge_column:
    discharge_data = discharge_data[discharge_column[0]].reset_index()
    discharge_data.columns = ['date', 'discharge']
    
    # Convert the 'date' column to datetime and remove timezone information
    discharge_data['date'] = pd.to_datetime(discharge_data['date']).dt.tz_localize(None)
    
    output_filename = 'discharge_data.xlsx'
    discharge_data.to_excel(output_filename, index=False)
    print(f"Discharge data has been successfully exported to '{output_filename}'.")
else:
    print("No discharge data found in the DataFrame.")
```

```python
import hydrofunctions as hf
import pandas as pd

def download_primary_discharge(site, start_date, end_date):
    """
    Download primary time series of discharge data for a specific site and time period from USGS.
    
    Parameters:
        site (str): USGS site number.
        start_date (str): Start date in 'YYYY-MM-DD' format.
        end_date (str): End date in 'YYYY-MM-DD' format.
        
    Returns:
        pandas.DataFrame: DataFrame containing primary time series of discharge data.
    """
    data = hf.NWIS(site, "iv", start_date, end_date)
    discharge = data.df('discharge')
    return discharge

# List of USGS site numbers
sites = ['08212400', '08212300', '08212820', '08211520', '08211500', '08189700', '08189500', '08188810', '08188590', '08164800', '08164600', '08464000', '08164450', '08164503', '08164504', '08162600', '08162501', '08189200']
start_date = '2001-01-01'
end_date = '2023-12-31'

for site in sites:
    # Download primary discharge data from USGS
    discharge_data = download_primary_discharge(site, start_date, end_date)

    # Convert datetime index to timezone-naive
    discharge_data.index = discharge_data.index.tz_localize(None)

    # Check if the DataFrame is not empty
    if not discharge_data.empty:
        # Export discharge data to Excel
        output_filename = f'discharge_data_{site}.xlsx'
        discharge_data.to_excel(output_filename, index=True, header=['Discharge'])
        print(f"Discharge data for site {site} has been successfully exported to '{output_filename}'.")
    else:
        print(f"No data available for site {site} for the specified time period.")
```

```python
import hydrofunctions as hf

def check_timezone(site):
    """
    Check the timezone information of the discharge data for a specific USGS site.
    
    Parameters:
        site (str): USGS site number.
        
    Returns:
        str: A string indicating the timezone information of the discharge data.
    """
    # Download primary discharge data from USGS
    data = hf.NWIS(site, "iv", "2000-01-01", "2000-01-02")
    discharge_data = data.df('discharge')

    # Check timezone information of the datetime index
    timezone_info = discharge_data.index.tzinfo

    if timezone_info is None:
        return "Timezone-naive"
    else:
        return str(timezone_info)

# Specify the USGS site number
site = '08164600'

# Check the timezone for the specified site
timezone = check_timezone(site)
print(f"Timezone information for site {site}: {timezone}")
```
