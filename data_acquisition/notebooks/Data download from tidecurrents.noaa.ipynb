```python
import requests
import pandas as pd
from datetime import datetime, timedelta
import io

# Function to generate date ranges
def generate_date_ranges(start_date, end_date, chunk_size=31):
    start = datetime.strptime(start_date, "%Y%m%d")
    end = datetime.strptime(end_date, "%Y%m%d")
    date_ranges = []
    while start < end:
        chunk_end = min(start + timedelta(days=chunk_size - 1), end)
        date_ranges.append((start.strftime("%Y%m%d"), chunk_end.strftime("%Y%m%d")))
        start = chunk_end + timedelta(days=1)
    return date_ranges

# NOAA Tides and Currents API URL for water level data
base_url = "https://api.tidesandcurrents.noaa.gov/api/prod/datagetter"

# Parameters for the API request (constant)
params = {
    'station': '8774230',       # Station ID for Port Lavaca, TX
    'product': 'water_level',   # Product type
    'datum': 'MSL',            # Datum (Mean Lower Low Water)
    'units': 'metric',          # Units (metric or english)
    'time_zone': 'gmt',         # Time zone (gmt or lst)
    'application': 'web_services', # Application name
    'format': 'csv'             # Format of the data (json or csv)
}

# Date range for the data download
start_date = '20220630'  # Start date in YYYYMMDD format
end_date = '20220730'    # End date in YYYYMMDD format

# Generate date ranges
date_ranges = generate_date_ranges(start_date, end_date)

# List to store dataframes
data_frames = []

# Download data for each date range
for begin_date, end_date in date_ranges:
    print(f"Downloading data from {begin_date} to {end_date}")
    params['begin_date'] = begin_date
    params['end_date'] = end_date
    response = requests.get(base_url, params=params)
    
    if response.status_code == 200:
        # Read the CSV content into a dataframe
        df = pd.read_csv(io.StringIO(response.text))
        data_frames.append(df)
    else:
        print(f"Error: {response.status_code}")
        print(response.text)

# Combine all dataframes into a single dataframe
if data_frames:
    combined_df = pd.concat(data_frames)
    # Save the combined dataframe to a CSV file
    combined_df.to_csv('Aransas_wildlife_refuge_8774230_july2022.csv', index=False)
    print("Data downloaded and saved to Aransas_wildlife_refuge_8774230_july2022.csv")
else:
    print("No data downloaded.")

```
