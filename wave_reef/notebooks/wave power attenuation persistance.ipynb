```python
import mikeio
import numpy as np
import pandas as pd

# =========================
# File paths
# =========================
dfsu_reef = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\Northside_barrier_reef_1.5m_v3.sw - Result Files\Northside_barrier_reef_1.5m_v3.dfsu"

dfsu_no_reef = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\Small_scale_Wave_Model_Cyclone_Harvey.sw - Result Files\small_scale_wave_result.dfsu"

excel_out = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\north_barrier_wave_power_all_timesteps_transect.xlsx"

# =========================
# Transect endpoints
# =========================
x1, y1 = 732144, 3178938
x2, y2 = 732762, 3177571

n_points = 250

x_line = np.linspace(x1, x2, n_points)
y_line = np.linspace(y1, y2, n_points)
coords = np.column_stack([x_line, y_line])

distance = np.sqrt(
    (x_line - x_line[0])**2 +
    (y_line - y_line[0])**2
)

# =========================
# Function: extract all time steps along transect
# =========================
def extract_wave_power_transect_all_times(dfsu_file, item_name="Wave power, p"):
    dfs = mikeio.open(dfsu_file)
    ds = dfs.read(items=[item_name])
    da = ds[item_name]

    elem_ids, weights = da.geometry.get_2d_interpolant(
        coords,
        n_nearest=3,
        extrapolate=False
    )

    all_profiles = []

    for t in range(len(ds.time)):
        da_t = da.isel(time=t)

        profile_t = da_t.geometry.interp2d(
            da_t,
            elem_ids,
            weights
        ).flatten()

        all_profiles.append(profile_t)

    all_profiles = np.array(all_profiles)

    point_cols = [f"P{i+1:03d}" for i in range(n_points)]

    df = pd.DataFrame(
        all_profiles,
        columns=point_cols
    )

    df.insert(0, "Time", pd.to_datetime(ds.time.values))

    df["Average_Wave_Power"] = df[point_cols].mean(axis=1, skipna=True)

    return df

# =========================
# Extract reef and no-reef wave power
# =========================
df_reef = extract_wave_power_transect_all_times(dfsu_reef)
df_no_reef = extract_wave_power_transect_all_times(dfsu_no_reef)

# =========================
# Optional: transect point information sheet
# =========================
df_points = pd.DataFrame({
    "Point_ID": [f"P{i+1:03d}" for i in range(n_points)],
    "Distance_along_transect_m": distance,
    "Easting_m": x_line,
    "Northing_m": y_line
})

# =========================
# Export to Excel
# =========================
with pd.ExcelWriter(excel_out, engine="openpyxl") as writer:
    df_reef.to_excel(writer, sheet_name="WavePower_With_Reef", index=False)
    df_no_reef.to_excel(writer, sheet_name="WavePower_No_Reef", index=False)
    df_points.to_excel(writer, sheet_name="Transect_Point_Info", index=False)

print("Excel file saved successfully:")
print(excel_out)
```
