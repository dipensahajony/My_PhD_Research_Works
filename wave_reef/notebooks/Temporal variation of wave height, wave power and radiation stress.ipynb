```python
import mikeio
import numpy as np
import matplotlib.pyplot as plt

# File paths
fringing_dfsu_path = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\West_side_fringing_reef.sw - Result Files\west_side_fringing_reef.dfsu"
barrier_dfsu_path = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\West_side_barrier_reef.sw - Result Files\small_scale_wave_result_with_barrier_reef.dfsu"

# Shore location
shore_xy = np.array([729721, 3174949])

def extract_swh_time_series(dfsu_path, location_xy):
    ds = mikeio.Dfsu(dfsu_path)
    elem_coords = np.array(ds.geometry.element_coordinates)[:, :2]  # (x, y) only

    # Find nearest element to the location
    distances = np.linalg.norm(elem_coords - location_xy, axis=1)
    nearest_idx = np.argmin(distances)

    # Read Significant Wave Height data
    data = ds.read(items="Sign. Wave Height")
    swh = data[0].values[:, nearest_idx]  # Time series for nearest element
    time = data.time

    return time, swh

# Extract time series for both cases
time_fringing, swh_fringing = extract_swh_time_series(fringing_dfsu_path, shore_xy)
time_barrier, swh_barrier = extract_swh_time_series(barrier_dfsu_path, shore_xy)
# Increase all text sizes
plt.rcParams.update({
    'font.size': 16,
    'axes.titlesize': 16,
    'axes.labelsize': 16,
    'xtick.labelsize': 14,
    'ytick.labelsize': 14
})

# Plotting
plt.figure(figsize=(10, 5))
plt.plot(time_fringing, swh_fringing, label="Fringing Reef", linewidth=2)
plt.plot(time_barrier, swh_barrier, label="Barrier Reef", linewidth=2)
plt.xlabel("Time")
plt.ylabel("Sign. Wave Height (m)")
plt.title("Temporal Variation of Significant Wave Height at Shore Location")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
