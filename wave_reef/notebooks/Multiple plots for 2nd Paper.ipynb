```python
import mikeio
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# File paths (use raw string `r""` to avoid escape character issues)
existing_file = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\Small_scale_Wave_Model_Cyclone_Harvey.sw - Result Files\small_scale_wave_result.dfsu"
altered_file = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\Small_scale_Wave_Model_Cyclone_Harvey_with_reef_barrier_reef_Pattern.sw - Result Files\small_scale_wave_result_with_barrier_reef.dfsu"

# Open DFSU files
dfs_existing = mikeio.Dfsu(existing_file)
dfs_altered = mikeio.Dfsu(altered_file)

# Check available items (prints names of stored variables)
print("Existing Bathymetry Items:", dfs_existing.items)
print("Altered Bathymetry Items:", dfs_altered.items)

# Find the correct index for "Significant wave height" (modify if necessary)
Hs_index = 0
for i, item in enumerate(dfs_existing.items):
    if "Significant wave height" in item.name.lower():  # Make sure name matches
        Hs_index = i
        break

if Hs_index is None:
    raise ValueError("Significant wave height not found in the DFSU file.")

# Read full dataset and extract significant wave height
Hs_existing = dfs_existing.read()[Hs_index].to_numpy().flatten()
Hs_altered = dfs_altered.read()[Hs_index].to_numpy().flatten()

# Create DataFrame for visualization
df = pd.DataFrame({
    "Condition": ["Existing"] * len(Hs_existing) + ["Altered"] * len(Hs_altered),
    "Significant Wave Height (m)": np.concatenate([Hs_existing, Hs_altered])
})

# Plot Boxplot
plt.figure(figsize=(7, 5))
sns.boxplot(x="Condition", y="Significant Wave Height (m)", hue="Condition", 
            data=df, palette={"Existing": "blue", "Altered": "red"}, legend=False)

plt.title("Comparison of Significant Wave Height for Different Bathymetries")
plt.xlabel("Bathymetry Condition")
plt.ylabel("Significant Wave Height (m)")
plt.grid(True, linestyle="--", alpha=0.5)

# Show plot
plt.show()
```

```python
import mikeio
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# File paths
existing_file = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\Small_scale_Wave_Model_Cyclone_Harvey.sw - Result Files\small_scale_wave_result.dfsu"
north_reef = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\Small_scale_Wave_Model_Cyclone_Harvey_with_reef_barrier_reef_Pattern.sw - Result Files\small_scale_wave_result_with_barrier_reef.dfsu"
west_reef = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\West_side_barrier_reef.sw - Result Files\small_scale_wave_result_with_barrier_reef.dfsu"
east_reef = r"D:\Phd Research\Wave Model\small scale port lavaca\Simulation\east_side_barrier_reef.sw - Result Files\east_side_reef.dfsu"

# Function to read Significant Wave Height (Hs) from a DFSU file
def read_wave_height(file_path):
    dfs = mikeio.Dfsu(file_path)
    
    # Find the index of "Significant Wave Height"
    Hs_index = 0
    for i, item in enumerate(dfs.items):
        if "significant wave height" in item.name.lower().strip():
            Hs_index = i
            break

    if Hs_index is None:
        raise ValueError(f"Error: 'Significant wave height' not found in {file_path}")

    # Read data and flatten array
    return dfs.read()[Hs_index].to_numpy().flatten()

# Read data from each scenario
data = []
for file_path, label in zip(
    [existing_file, north_reef, west_reef, east_reef], 
    ["Existing", "North Reef", "West Reef", "East Reef"]
):
    Hs_values = read_wave_height(file_path)
    data.extend(zip([label] * len(Hs_values), Hs_values))  # Store as tuples

# Convert to DataFrame
df = pd.DataFrame(data, columns=["Condition", "Significant Wave Height (m)"])
# Increase all default font sizes
plt.rcParams.update({
    'font.size': 14,         # base font size for text
    'axes.titlesize': 14,    # title
    'axes.labelsize': 14,    # x/y labels
    'xtick.labelsize': 14,   # x tick labels
    'ytick.labelsize': 14,   # y tick labels
    'legend.fontsize': 14    # legend
})
# Plot Boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(x="Condition", y="Significant Wave Height (m)", data=df, 
            palette={"Existing": "blue", "North Reef": "green", 
                     "West Reef": "orange", "East Reef": "red"})

plt.title("Comparison of Significant Wave Height for Different Scenarios")
plt.xlabel("Scenario")
plt.ylabel("Significant Wave Height (m)")
plt.grid(True, linestyle="--", alpha=0.5)

# Show plot
plt.show()
```
