```python
import mikeio
import numpy as np

# Load DFSU file
dfsu = mikeio.Dfsu("D:\Phd Research\Wave Model\Simulation\Wave_Model_Cyclone_Harvey.sw - Result Files\wave_result.dfsu")

# Get all available items
print(dfsu.items)  # Check for 'Significant wave height' and 'Peak wave period'

# Define coordinates of the target location (or use an index)
target_x, target_y = 733629, 3177484  # Replace with your location

# Find the nearest element index
element_index = dfsu.find_nearest_element((target_x, target_y))

# Extract Hs and Tp at this location
Hs = dfsu.read(["Significant wave height"]).to_numpy()[:, element_index]
Tp = dfsu.read(["Peak wave period"]).to_numpy()[:, element_index]  # Adjust if using Tm

# Define period threshold (common choice: 10-12 s for Tp)
threshold = 10

# Compute SS and IG wave height
Hs_ss = np.where(Tp > threshold, Hs, 0)  # SS waves (Tp > 10s)
Hs_ig = np.where(Tp <= threshold, Hs, 0) # IG waves (Tp <= 10s)

# Print results
print(f"Location: ({target_x}, {target_y})")
print("Total Hs:", Hs)
print("Sea-Swell Hs (Hs_ss):", Hs_ss)
print("Infragravity Hs (Hs_ig):", Hs_ig)
```
