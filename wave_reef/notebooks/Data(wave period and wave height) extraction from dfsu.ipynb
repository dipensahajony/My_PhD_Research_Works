```python
import numpy as np
import pandas as pd
import mikeio

# ============================================================
# INPUTS
# ============================================================
dfsu_file = r"D:\Phd Research\Wave Model\Simulation\Wave_Model_Sep_2022.sw - Result Files\wave_result_v5.dfsu"

target_wave_heights = [
    [0.016, 0.047, 0.397],
    [0.004, 0.0105, 0.347],
    [0.004, 0.01105, 0.286],
    [0.013, 0.051, 0.287],
    [0.003, 0.009, 0.268],
]

# ============================================================
# READ DFSU
# ============================================================
dfs = mikeio.open(dfsu_file)

# Read only required items
ds = dfs.read(items=["Sign. Wave Height", "Wave Period, T02"])

hs_data = ds["Sign. Wave Height"].to_numpy()
t02_data = ds["Wave Period, T02"].to_numpy()

# Make sure arrays are 2D: (time, element)
if hs_data.ndim == 1:
    hs_data = hs_data[np.newaxis, :]
if t02_data.ndim == 1:
    t02_data = t02_data[np.newaxis, :]

times = ds.time
n_time, n_elem = hs_data.shape

# ============================================================
# FLATTEN DATA
# ============================================================
hs_flat = hs_data.flatten()
t02_flat = t02_data.flatten()

valid_mask = np.isfinite(hs_flat) & np.isfinite(t02_flat)

hs_valid = hs_flat[valid_mask]
t02_valid = t02_flat[valid_mask]
valid_indices = np.where(valid_mask)[0]

# ============================================================
# FIND CLOSEST T02 FOR EACH TARGET Hs
# ============================================================
results = []
t02_matrix = []

for i, row in enumerate(target_wave_heights, start=1):
    t02_row = []
    for j, target_hs in enumerate(row, start=1):

        diff = np.abs(hs_valid - target_hs)
        idx = np.argmin(diff)

        matched_hs = hs_valid[idx]
        matched_t02 = t02_valid[idx]
        error = diff[idx]

        flat_index = valid_indices[idx]
        time_index = flat_index // n_elem
        elem_index = flat_index % n_elem

        t02_row.append(matched_t02)

        results.append({
            "row": i,
            "col": j,
            "target_Hs": target_hs,
            "matched_Hs": matched_hs,
            "corresponding_T02": matched_t02,
            "abs_error": error,
            "time": str(times[time_index]),
            "element_index": elem_index
        })

    t02_matrix.append(t02_row)

# ============================================================
# PRINT DETAILED RESULTS
# ============================================================
df_results = pd.DataFrame(results)

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
pd.set_option("display.float_format", lambda x: f"{x:.6f}")

print("\nDetailed matching results:")
print(df_results.to_string(index=False))

# ============================================================
# PRINT ONLY THE T02 VALUES IN SAME 5x3 FORMAT
# ============================================================
print("\nCorresponding T02 values in same 5x3 format:")
for row in t02_matrix:
    print("  ".join(f"{v:.6f}" for v in row))
```
