```python
import matplotlib.pyplot as plt
import numpy as np

# Criteria and normalized scores
criteria = [
    "Wave Energy Reduction", "Wave Directionality", "Wind Directionality",
    "Energy Dissipation", "Runup Reduction", "Exposure",
    "Vulnerability", "Historical Inundation"
]

# Normalized scores per location (values from your MCA analysis)
scores = {
    "North": [0.6, 0.5, 0.4, 0.5, 0.9, 0.4, 0.3, 0.3],
    "West":  [0.8, 0.8, 0.9, 0.7, 0.8, 0.9, 0.9, 0.7],
    "East":  [0.7, 0.6, 0.5, 0.6, 0.7, 0.6, 0.8, 0.6]
}

# Setup
labels = np.array(criteria)
num_vars = len(labels)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]  # complete the loop

# Function to plot each location
def add_to_radar(ax, values, label, color):
    values += values[:1]
    ax.plot(angles, values, label=label, linewidth=2, color=color)
    ax.fill(angles, values, alpha=0.25, color=color)
# Increase all text sizes
plt.rcParams.update({
    'font.size': 16,
    'axes.titlesize': 18,
    'axes.labelsize': 16,
    'xtick.labelsize': 16,
    'ytick.labelsize': 16
})

# Create figure
fig, ax = plt.subplots(figsize=(10, 10), subplot_kw=dict(polar=True))
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)

# Draw axis
ax.set_thetagrids(np.degrees(angles[:-1]), labels)
ax.set_ylim(0, 1)

# Plot all locations
colors = ["#1f77b4", "#2ca02c", "#ff7f0e"]  # blue, green, orange
for i, (loc, vals) in enumerate(scores.items()):
    add_to_radar(ax, vals.copy(), loc, colors[i])

# Legend and title
ax.legend(loc="upper right", bbox_to_anchor=(1.3, 1.1))
plt.title("Multi-Criteria Reef Suitability Comparison", size=14)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# Data
# ==========================================================
locations = ["North", "West", "East"]

hydrodynamic_score = np.array([0.36, 0.42, 0.75])
environmental_score = np.array([0.60, 0.73, 0.26])

# ==========================================================
# Decision-space reference lines
# Mean values are used as the quadrant boundaries
# ==========================================================
hydro_mean = hydrodynamic_score.mean()
environment_mean = environmental_score.mean()

# ==========================================================
# Create figure
# ==========================================================
fig, ax = plt.subplots(figsize=(6,4), dpi=300)

# Scatter points
ax.scatter(
    hydrodynamic_score,
    environmental_score,
    s=90,
    edgecolor="black",
    linewidth=0.7,
    zorder=3
)

# ==========================================================
# Point labels
# ==========================================================
label_offsets = {
    "North": (0.012, 0.012),
    "West": (0.012, 0.012),
    "East": (0.012, 0.012)
}

for location, x, y in zip(
    locations,
    hydrodynamic_score,
    environmental_score
):
    dx, dy = label_offsets[location]

    ax.text(
        x + dx,
        y + dy,
        location,
        fontsize=11,
        ha="left",
        va="bottom"
    )

# ==========================================================
# Quadrant reference lines
# ==========================================================
ax.axvline(
    hydro_mean,
    linestyle="--",
    linewidth=1.4,
    color="gray"
)

ax.axhline(
    environment_mean,
    linestyle="--",
    linewidth=1.4,
    color="gray"
)

# ==========================================================
# Axis formatting
# ==========================================================
# ax.set_xlabel(
#     "Hydrodynamic score",
#     fontsize=12,
#     fontweight="bold"
# )

# ax.set_ylabel(
#     "Environmental score",
#     fontsize=12,
#     fontweight="bold"
# )

# ax.set_title(
#     "(c) 2-D decision space",
#     fontsize=14,
#     pad=10
# )

ax.set_xlim(0.0, 1.0)
ax.set_ylim(0.0, 1.0)

# ax.set_xticks(np.arange(0.2, 0.81, 0.1))
# ax.set_yticks(np.arange(0.2, 0.86, 0.1))

ax.tick_params(
    axis="both",
    labelsize=14,
    direction="out",
    width=1
)

# Border formatting
for spine in ax.spines.values():
    spine.set_linewidth(1.1)

# No regular grid, only the two decision-boundary lines
ax.grid(False)

plt.tight_layout()

# # Optional high-resolution export
# plt.savefig(
#     "2D_decision_space.png",
#     dpi=600,
#     bbox_inches="tight"
# )

plt.show()

# ==========================================================
# Print boundary values
# ==========================================================
# print(f"Mean hydrodynamic score: {hydro_mean:.3f}")
# print(f"Mean environmental score: {environment_mean:.3f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ==========================
# Data
# ==========================
locations = ["North", "West", "East"]
mca_score = [0.48, 0.57, 0.50]

# ==========================
# Plot
# ==========================
plt.figure(figsize=(6, 4), dpi=300)

bars = plt.bar(
    locations,
    mca_score,
    width=0.55,
    color=["#4C72B0", "#DD8452", "#55A868"],
    edgecolor="black",
    linewidth=1
)

# ==========================
# Labels
# ==========================
# plt.xlabel("Reef Location", fontsize=13, fontweight="bold")
# plt.ylabel("MCA Score", fontsize=13, fontweight="bold")

plt.xticks(fontsize=16)
plt.yticks(np.arange(0, 0.81, 0.1), fontsize=12)

plt.ylim(0, 0.8)

# Horizontal grid
plt.grid(axis="y", linestyle="--", alpha=0.4)

# ==========================
# Value labels
# ==========================
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 0.015,
        f"{height:.2f}",
        ha="center",
        va="bottom",
        fontsize=11,
        fontweight="bold"
    )

# Thicker axes
ax = plt.gca()
for spine in ax.spines.values():
    spine.set_linewidth(1.2)

plt.tight_layout()

# # Save publication-quality figure
# plt.savefig("MCA_scores_bar_chart.png", dpi=600, bbox_inches="tight")

plt.show()
```
