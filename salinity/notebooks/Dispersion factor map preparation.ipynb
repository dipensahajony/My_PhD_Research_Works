```python
import mikeio
from mikeio import Mesh, Dataset, GeometryFM2D, ItemInfo, EUMUnit
import numpy as np
from mikeio import EUMType

# Mesh file path
mesh_file = r"D:\Phd Research\Prototype HD Model\Mesh File\Prototype_domain_without_river_network_utm - Copy.mesh"
mesh = Mesh(mesh_file)

# Define dispersion coefficient values (default: 5 m²/s)
dispersion_values = np.full(mesh.n_elements, 5.0)

# Adjust values based on conditions (e.g., depth)
for i, (x, y, z) in enumerate(mesh.element_coordinates):
    if z < 1.0:
        dispersion_values[i] = 0.5

# Output DFSU file path
output_file = r"D:\Phd Research\Prototype HD Model\Mesh File\texas_dispersion_map.dfsu"

# Create Geometry
geometry = GeometryFM2D(
    node_coordinates=mesh.node_coordinates,
    element_table=mesh.element_table,
    projection="UTM-14"
)

# Create Dataset object
dataset = Dataset(
    data=[dispersion_values],
    items=[ItemInfo("Dispersion coefficient", EUMType.Dispersion_coefficient, EUMUnit.square_meter_per_sec)],
    geometry=geometry,
)

# Write the DFSU file
dataset.to_dfs(output_file)

print(f"Successfully created DFSU file at: {output_file}")
```
