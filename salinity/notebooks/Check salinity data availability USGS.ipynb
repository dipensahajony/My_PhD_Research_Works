```python
import pandas as pd
from dataretrieval import nwis
import matplotlib.pyplot as plt

# USGS site and parameter info
site = "08162501"           # Colorado River at Bay City, TX
parameter_code = "00480"    # Salinity (ppt)
start = "2020-01-01"
end = "2024-12-31"

# Try to get daily salinity data
df = nwis.get_record(sites=site, parameterCd=parameter_code, start=start, end=end, service='dv')

# Check and plot
if df.empty:
    print("❌ No daily salinity data found for site", site)
    print("🔁 Trying to fetch instantaneous data instead...")

    # Try instantaneous (real-time) data
    df_rt = nwis.get_record(sites=site, parameterCd=parameter_code, start=start, end=end, service='iv')

    if df_rt.empty:
        print("❌ No salinity data available (daily or real-time) for this station and time range.")
    else:
        print("✅ Instantaneous salinity data retrieved!")
        df_rt['00480'] = pd.to_numeric(df_rt['00480'], errors='coerce')
        df_rt['00480'].plot(figsize=(12, 6), title="Instantaneous Salinity at USGS 08164600", ylabel="Salinity (ppt)")
        plt.grid(True)
        plt.tight_layout()
        plt.show()
else:
    print("✅ Daily salinity data retrieved!")
    df['00480_Mean'] = pd.to_numeric(df['00480_Mean'], errors='coerce')
    df['00480_Mean'].plot(figsize=(12, 6), title="Daily Salinity at USGS 08164600", ylabel="Salinity (ppt)")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

```
