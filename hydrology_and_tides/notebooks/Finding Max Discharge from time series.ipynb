```python
import pandas as pd
import os

# Folder path where the Excel files are stored
folder_path = r"D:\Phd Research\Data\Discharge Data\Raw Data"

# List of Excel files in the folder (adjust extension if necessary)
excel_files = [f for f in os.listdir(folder_path) if f.endswith('.xlsx')]

# Create a Pandas ExcelWriter object to write to one file
output_filepath = r"D:\Phd Research\Data\Discharge Data\Output\max_discharge_per_year.xlsx"
with pd.ExcelWriter(output_filepath, engine='xlsxwriter') as writer:
    
    # Loop through each file
    for file in excel_files:
        file_path = os.path.join(folder_path, file)
        
        # Load the Excel file (adjust sheet name if needed)
        df = pd.read_excel(file_path, sheet_name='Sheet2')
        
        # Ensure 'Date' is in datetime format
        df['Date'] = pd.to_datetime(df['Date'])
        
        # Extract the year from 'Date'
        df['Year'] = df['Date'].dt.year
        
        # Find the maximum discharge and corresponding date for each year
        max_discharge_per_year = df.loc[df.groupby('Year')['Discharge (m3/s)'].idxmax()]
        
        # Select only the 'Date' and 'Discharge (m3/s)' columns
        result = max_discharge_per_year[['Date', 'Discharge (m3/s)']]
        
        # Write to Excel in a new sheet named after the file
        sheet_name = os.path.splitext(file)[0]  # Use filename as sheet name
        result.to_excel(writer, sheet_name=sheet_name, index=False)

# The output file is now saved at 'output_filepath' with multiple sheets.
```
