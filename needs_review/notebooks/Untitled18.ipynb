```python
import rasterio
import matplotlib.pyplot as plt
import numpy as np

# === File path to flood depth difference raster ===
raster_path = r"D:\Phd Research\GIS\Raster\inund_diff_SLR_and_50yr_simulated.tif"

# === Load the raster ===
with rasterio.open(raster_path) as src:
    data = src.read(1)
    data[data == src.nodata] = np.nan  # mask nodata
    bounds = src.bounds
    extent = [bounds.left, bounds.right, bounds.bottom, bounds.top]

# === Plot the raster ===
plt.figure(figsize=(10, 8))
im = plt.imshow(data, cmap='viridis', extent=extent, origin='upper', vmin=0, vmax=10)
cbar = plt.colorbar(im, fraction=0.036, pad=0.04)
cbar.set_label("Flood Depth Difference (m)", fontsize=12)

plt.title("Flood Depth Difference: SLR vs 50-Year Simulated", fontsize=14)
plt.xlabel("Easting (m)")
plt.ylabel("Northing (m)")
plt.grid(False)
plt.tight_layout()
plt.show()
```
