```python
import mikeio
import numpy as np
import pandas as pd

# Load DFSU file
dfsu = mikeio.Dfsu("D:/Phd Research/Wave Model/Simulation/Wave_Model_Cyclone_Harvey.sw - Result Files/wave_result.dfsu")

# Define target coordinates
target_x, target_y = 733629, 3177484  

# Extract node coordinates
node_x, node_y = dfsu.geometry.node_coordinates[:, 0], dfsu.geometry.node_coordinates[:, 1]

# Compute Euclidean distance to target location
distances = np.sqrt((node_x - target_x)**2 + (node_y - target_y)**2)

# Get the nearest node index
nearest_node_index = np.argmin(distances)

# Read all data
data = dfsu.read()

# Extract Hs and Tp at this node
Hs = data["Sign. Wave Height"].to_numpy()[:, nearest_node_index]
Tp = data["Wave Period, T02"].to_numpy()[:, nearest_node_index]

# Define period threshold (e.g., 10s for IG vs. SS separation)
threshold = 10

# Compute SS and IG wave height
Hs_ss = np.where(Tp > threshold, Hs, 0)  # SS waves (Tp > 10s)
Hs_ig = np.where(Tp <= threshold, Hs, 0) # IG waves (Tp <= 10s)

# Get time steps
time_steps = data.time

# Create a DataFrame
df = pd.DataFrame({
    "Time": time_steps,
    "Total Hs (m)": Hs,
    "Sea-Swell Hs (m)": Hs_ss,
    "Infragravity Hs (m)": Hs_ig,
    "Wave Period (s)": Tp
})

# Export to Excel
output_path = "D:/Phd Research/Wave Model/Simulation/wave_height_output.xlsx"
df.to_excel(output_path, index=False)

print(f"Data exported successfully to {output_path}")
```
