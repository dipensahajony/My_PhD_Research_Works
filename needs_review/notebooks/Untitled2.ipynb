```python
import pandas as pd
from openpyxl import load_workbook
# Load the Excel file
file_path = r'D:\Phd Research\1D HD Model\xyz 6 rivers.xlsx'
wb = load_workbook(file_path)
sheet = wb.active

 #Read data from the Excel sheet into a DataFrame
data = pd.DataFrame(sheet.values)
header = data.iloc[0]
data = data[1:]
data.columns = header

#Sort the data by 'RvName' and 'ET_ID'
data.sort_values(by=['RvName', 'ET_ID'], inplace=True)

# Initialize variables for tracking chainage and CS count
chainage = 0
cs_count = 1

# Lists to store chainage and CS values
chainage_values = []
cs_values = []

# Iterate through each row
for index, row in data.iterrows():
    # If it's a new 'Rv name', reset chainage to 0 and CS count to 1
    if index == 0 or row['RvName'] != data.at[index - 1, 'RvName']:
        chainage = 0
        cs_count = 1
    else:
        # Increment chainage if 'Rv name' is the same as the previous row
        chainage += 1
    
    # Append chainage and CS values to the lists
    chainage_values.append(chainage)
    cs_values.append('CS-' + str(cs_count))
    
    # Increment CS count
    cs_count += 1 if row['ET_ID'] != data.at[index + 1, 'ET_ID'] else 0

# Add the chainage and CS values to the DataFrame
data['Chainage'] = chainage_values
data['CS'] = cs_values

# Save the updated DataFrame to a new sheet in the same Excel file
with pd.ExcelWriter(file_path, engine='openpyxl') as writer:
    writer.book = wb
    data.to_excel(writer, index=False, sheet_name='Updated_Data')

# Save the changes to the Excel file
wb.save(file_path)
```
