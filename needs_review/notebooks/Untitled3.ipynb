```python
import mikeio
import matplotlib.pyplot as plt

# Replace 'your_file.dfs0' with the path to your DFS0 file
dfs = mikeio.open(r'D:\Phd Research\Prototype HD Model\Measured WL\Aransas wildlife refuge JUL 2023.dfs0')

# Read the data from the DFS0 file
data = dfs.read()

# Assuming the first item in the file is the one you want to plot
time_series = data[0]

# Plotting the time series data
plt.figure(figsize=(10, 6))
plt.plot(time_series.time, time_series.data, label=dfs.items[0].name)
plt.xlabel('Time')
plt.ylabel(dfs.items[0].name)
plt.title('DFS0 Data Plot')
plt.legend()
plt.grid(True)
plt.show()
```
