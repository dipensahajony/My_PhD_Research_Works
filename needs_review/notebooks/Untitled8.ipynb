```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np

# ---------------------- COMMON VALUE BOUNDS ----------------------
bounds = [0, 0.1, 1, 2.5, 5, 7.5, 10, 13.786]
labels = ['0', '0.1', '1', '2.5', '5', '7.5', '≥10']

# ---------------------- DEFINE COLOR RAMPS ----------------------
ramps = {
    "Warm (Yellow–Red)": [
        "#ffff00", "#ffcc00", "#ff9900", "#ff6600", "#ff3300", "#ff0000", "#b30000"
    ],
    "Cool (Blue–Cyan)": [
        "#0000ff", "#0066ff", "#0099ff", "#00ccff", "#00ffff", "#33ffff", "#66ffff"
    ],
    "Violet (Purple–Magenta)": [
        "#8000ff", "#9933ff", "#b366ff", "#cc66ff", "#ff66ff", "#ff33cc", "#ff0099"
    ],
}

# ---------------------- PLOT COLORBARS ----------------------
fig, axes = plt.subplots(nrows=len(ramps), figsize=(8, 3.8))
fig.subplots_adjust(left=0.1, right=0.95, top=0.95, bottom=0.25, hspace=0.8)

for ax, (title, colors) in zip(axes, ramps.items()):
    cmap = mpl.colors.LinearSegmentedColormap.from_list(
        f"{title}_cmap", list(zip(np.linspace(0, 1, len(colors)), colors))
    )
    norm = mpl.colors.Normalize(vmin=min(bounds), vmax=max(bounds))
    
    cb = mpl.colorbar.ColorbarBase(
        ax,
        cmap=cmap,
        norm=norm,
        orientation='horizontal'
    )
    
    cb.set_ticks(bounds[:-1])
    cb.ax.set_xticklabels(labels, fontsize=10)
    cb.set_label(f"{title}  |  Flood Depth (m)", fontsize=11, weight='bold')
    cb.ax.tick_params(length=0)
    
plt.show()
```
