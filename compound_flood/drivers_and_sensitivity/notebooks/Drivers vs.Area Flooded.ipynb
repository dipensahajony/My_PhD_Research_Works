```python
import matplotlib.pyplot as plt
from matplotlib.ticker import StrMethodFormatter

# Data
scenarios = [
    "Only SLR",
    "Only Storm Surge",
    "Only River Flow",
    "Surge + Q",
    "Surge + SLR",
    "SLR + Q",
    "Compound",
]
area_flooded = [1877, 13117, 2255, 13175, 13450, 3397, 13583]  # km²

# Colors (singles, pairwise combos, compound)
colors = [
    "skyblue",          # Only SLR
    "orange",           # Only Storm Surge
    "green",            # Only River Flow
    "#9370DB",          # Surge + Q (purple)
    "#8A2BE2",          # Surge + SLR (blueviolet)
    "#BA55D3",          # SLR + Q (mediumorchid)
    "red",              # Compound
]

plt.figure(figsize=(8, 8))
bars = plt.bar(scenarios, area_flooded, color=colors, edgecolor="black", linewidth=0.5)

# Y-limit so labels fit above bars
plt.ylim(0, max(area_flooded) * 1.12)

# Add labels above bars with thousands separators
for bar, val in zip(bars, area_flooded):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height() + max(area_flooded)*0.02,  # small offset
        f"{val:,.0f}",
        ha="center", va="bottom", fontsize=13
    )

# Title and labels
plt.title("Drivers vs. Area Flooded", fontsize=18, weight="bold")
plt.ylabel("Area Flooded (km²)", fontsize=16)
plt.xlabel("Flood Drivers", fontsize=16)

# Tidy axes
plt.gca().yaxis.set_major_formatter(StrMethodFormatter('{x:,.0f}'))
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.xticks(rotation=20, ha="right")

plt.tight_layout()
plt.show()
```
