```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt

# ---------------- USER INPUT ----------------
raster_path = r"D:\Phd Research\sim_depth_100yr.tif"
bin_width = 0.3   # meters, adjust as needed
x_max = 12.0      # maximum depth to show on x-axis
# --------------------------------------------

# Open raster
with rasterio.open(raster_path) as src:
    arr = src.read(1).astype(float)
    nodata = src.nodata
    # Pixel size (assume square pixels)
    res_x, res_y = src.res
    pixel_area_m2 = abs(res_x * res_y)
    pixel_area_km2 = pixel_area_m2 / 1e6

# Mask nodata and invalid values
if nodata is not None:
    valid = (arr != nodata) & np.isfinite(arr)
else:
    valid = np.isfinite(arr)

vals = arr[valid]
vals = vals[vals > 0]   # keep only flooded pixels (>0 m)

# Bins from 0 to 12 m
bins = np.arange(0, x_max + bin_width, bin_width)

# Histogram weighted by pixel area (so y-axis = km²)
hist, edges = np.histogram(vals, bins=bins)
area_km2 = hist * pixel_area_km2

# Plot
plt.figure(figsize=(9,5))
plt.bar(edges[:-1], area_km2, width=bin_width, align="edge", edgecolor="black")
plt.xlim(0, x_max)
plt.xlabel("Water depth (m)")
plt.ylabel("Flooded area (km²)")
plt.title("Flooded area distribution by depth (100-year event)")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()

# # Print summary
# for d0, d1, a in zip(edges[:-1], edges[1:], area_km2):
#     if a > 0:
#         print(f"Depth {d0:.1f}–{d1:.1f} m: {a:.3f} km²")
# print(f"Total flooded area (≤{x_max} m): {area_km2.sum():.3f} km²")
```

```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import StrMethodFormatter

# ---------------- USER INPUT ----------------
raster_path = r"D:\Phd Research\Final_Raster\100yr_compound_flood_base_stat.tif"
bin_width = 0.3   # meters
x_max = 12.0      # maximum depth to show on x-axis
ref_lines = [0.1, 1.0, 3.0, 6.0, 7.0]  # vertical reference lines (m)
# --------------------------------------------

# Open raster
with rasterio.open(raster_path) as src:
    arr = src.read(1).astype(float)
    nodata = src.nodata

# Mask nodata and invalid values; keep only flooded (>0 m)
if nodata is not None:
    valid = (arr != nodata) & np.isfinite(arr)
else:
    valid = np.isfinite(arr)

vals = arr[valid]
vals = vals[vals > 0]

# Depth bins and histogram (COUNTS of cells per depth bin)
bins = np.arange(0, x_max + bin_width, bin_width)
counts, edges = np.histogram(vals, bins=bins)

# Plot (counts vs depth bins)
fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(edges[:-1], counts, width=bin_width, align="edge", edgecolor="black")
ax.set_xlim(0, x_max)
ax.set_xlabel("Water depth (m)")
ax.set_ylabel("Number of grid cells")
ax.set_title("Frequency distribution by flooded grid cell by depth")

# Format y axis with thousands separators (optional)
ax.yaxis.set_major_formatter(StrMethodFormatter('{x:,.0f}'))

# Add reference vertical lines
for x in ref_lines:
    if 0 <= x <= x_max:
        ax.axvline(x, linestyle="--", color="red", linewidth=1)
        # small label above the line (optional)
        ax.text(x, ax.get_ylim()[1]*0.98, f"{x:g} m",
                color="red", ha="center", va="top", fontsize=9, rotation=0)

ax.grid(axis="y", linestyle="--", alpha=0.6)
fig.tight_layout()
plt.show()

```
