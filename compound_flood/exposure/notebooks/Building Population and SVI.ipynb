```python
import pandas as pd
import geopandas as gpd
import numpy as np

# Load the shapefile for the land boundary
shapefile_path = r"D:\Phd Research\GIS\Shape\land_Part_area_utm.zip"
land_boundary = gpd.read_file(shapefile_path)

# Load the geodatabases and reproject them to the desired CRS
gdb1 = r"D:\Phd Research\Critical Infrastructure\Shape file\Building_Population_SVI\FPR10Bldgs_SVI_Pop_Nov2021\FPR10Bldgs_SVI_Pop_Nov2021.gdb"
gdb2 = r"D:\Phd Research\Critical Infrastructure\Shape file\Building_Population_SVI\FPR11Bldgs_SVI_Pop_Nov2021\FPR11Bldgs_SVI_Pop_Nov2021.gdb"
gdb3 = r"D:\Phd Research\Critical Infrastructure\Shape file\Building_Population_SVI\FPR12Bldgs_SVI_Pop_Nov2021\FPR12Bldgs_SVI_Pop_Nov2021.gdb"
gdb4 = r"D:\Phd Research\Critical Infrastructure\Shape file\Building_Population_SVI\FPR13Bldgs_SVI_Pop_Nov2021\FPR13Bldgs_SVI_Pop_Nov2021.gdb"
gdb5 = r"D:\Phd Research\Critical Infrastructure\Shape file\Building_Population_SVI\FPR14Bldgs_SVI_Pop_Nov2021\FPR14Bldgs_SVI_Pop_Nov2021.gdb"
gdb6 = r"D:\Phd Research\Critical Infrastructure\Shape file\Building_Population_SVI\FPR15Bldgs_SVI_Pop_Nov2021\FPR15Bldgs_SVI_Pop_Nov2021.gdb"

# Read and reproject the GeoDataFrames
fpr10 = gpd.read_file(gdb1).to_crs(32614)
fpr11 = gpd.read_file(gdb2).to_crs(32614)
fpr12 = gpd.read_file(gdb3).to_crs(32614)
fpr13 = gpd.read_file(gdb4).to_crs(32614)
fpr14 = gpd.read_file(gdb5).to_crs(32614)
fpr15 = gpd.read_file(gdb6).to_crs(32614)

# Perform intersection of each FPR layer with the land boundary
fpr10_intersect = gpd.overlay(fpr10, land_boundary, how='intersection')
fpr11_intersect = gpd.overlay(fpr11, land_boundary, how='intersection')
fpr12_intersect = gpd.overlay(fpr12, land_boundary, how='intersection')
fpr13_intersect = gpd.overlay(fpr13, land_boundary, how='intersection')
fpr14_intersect = gpd.overlay(fpr14, land_boundary, how='intersection')
fpr15_intersect = gpd.overlay(fpr15, land_boundary, how='intersection')

# Concatenate the results into one GeoDataFrame
fpr_combined = gpd.GeoDataFrame(pd.concat([fpr10_intersect, fpr11_intersect, fpr12_intersect, fpr13_intersect, fpr14_intersect, fpr15_intersect], ignore_index=True))

# Plot the combined GeoDataFrame
fpr_combined.plot()
```

```python
# Export fpr_combined as a shapefile
output_shapefile = r"D:\Phd Research\GIS\Shape\fpr_combined_intersection.shp"
fpr_combined.to_file(output_shapefile, driver='ESRI Shapefile')
```
