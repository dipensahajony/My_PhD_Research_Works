```python
import geopandas as gpd
import numpy as np
from shapely.geometry import Point, Polygon
import matplotlib.pyplot as plt
```

```python
# Load the shapefile for the land boundary
shapefile_path = r"D:\Phd Research\GIS\Shape\land_Part_area_utm.zip"
shape_path2 = r"D:\Phd Research\Critical Infrastructure\Shape file\Schools2019to2020_372503062685561240.zip"
land_boundary = gpd.read_file(shapefile_path)
school_point = gpd.read_file(shape_path2).to_crs(32614)
school_intersect = gpd.overlay(school_point, land_boundary, how='intersection')

# Count occurrences of each school type in the 'Grade_Leve' column
school_counts = school_intersect['Grade_Leve'].value_counts()
# Calculate percentages
percentages = 100 * school_counts / school_counts.sum()

# Create labels with both the school type and the percentage
labels = [f'{school_type}: {percentage:.1f}%' for school_type, percentage in zip(school_counts.index, percentages)]

# Plot the pie chart
plt.figure(figsize=(6, 6))
wedges, texts, autotexts = plt.pie(school_counts, labels=None, autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)

# Add a legend with school types and percentages
plt.legend(wedges, labels, title="School Types", loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))

plt.title('Schools 171 nos.')
plt.axis('equal')  # Equal aspect ratio ensures the pie chart is circular.

# Show the plot
plt.show()
```
