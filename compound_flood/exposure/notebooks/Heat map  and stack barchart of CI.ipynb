```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Data preparation
data = {
    'Return Period': ['5 Yr', '5 Yr', '5 Yr', '10 Yr', '10 Yr', '10 Yr', '20 Yr', '20 Yr', '20 Yr', '50 Yr', '50 Yr', '50 Yr'],
    'Approach': ['Current Approach', 'Only SLR (Bathtub)', 'Only Surge Level'] * 4,
    'School (Flood Affected)': [86, 0, 106, 138, 0, 132, 153, 0, 147, 168, 0, 158],
    'School (Flood Safe)': [85, 171, 65, 33, 171, 39, 18, 171, 24, 3, 171, 13],
    'Hospital (Flood Affected)': [9, 0, 6, 11, 0, 10, 12, 0, 12, 19, 0, 17],
    'Hospital (Flood Safe)': [13, 22, 16, 11, 22, 12, 10, 22, 10, 3, 22, 5],
    'Fire and EMS (Flood Affected)': [34, 0, 31, 47, 0, 45, 53, 0, 47, 58, 0, 55],
    'Fire and EMS (Flood Safe)': [28, 62, 31, 15, 62, 17, 9, 62, 15, 4, 62, 7],
    'National Shelter (Flood Affected)': [16, 0, 26, 37, 0, 36, 42, 0, 39, 43, 0, 41],
    'National Shelter (Flood Safe)': [31, 47, 21, 10, 47, 11, 5, 47, 8, 4, 47, 6],
}

df = pd.DataFrame(data)

# Ensure proper order for Return Periods
df['Return Period'] = pd.Categorical(df['Return Period'], categories=['5 Yr', '10 Yr', '20 Yr', '50 Yr'], ordered=True)

# Reshape the data for heatmap
heatmap_data = df.melt(
    id_vars=['Return Period', 'Approach'],
    var_name='Infrastructure Type',
    value_name='Count'
)

# Pivot table for heatmap
heatmap_pivot = heatmap_data.pivot_table(
    index='Infrastructure Type', columns=['Return Period', 'Approach'], values='Count'
)

# Plot the heatmap
plt.figure(figsize=(20, 14))
sns.heatmap(
    heatmap_pivot,
    cmap='coolwarm',
    annot=True,
    fmt='g',
    linewidths=0.5,
    cbar_kws={'label': 'Count'},
    annot_kws={"size": 14}  # Increase annotation font size
)

# Adjust font sizes and layout
plt.title('Flood Affected and Flood Safe Infrastructure', fontsize=24, weight='bold')
plt.xticks(fontsize=16, rotation=45, ha='right')
plt.yticks(fontsize=16, rotation=0)
plt.xlabel('Return Period and Approach', fontsize=18, weight='bold')
plt.ylabel('Infrastructure Type', fontsize=18, weight='bold')
cbar = plt.gca().collections[0].colorbar
cbar.ax.tick_params(labelsize=16)  # Increase colorbar font size
cbar.set_label('Count', fontsize=18, weight='bold')

plt.tight_layout()

# Show the heatmap
plt.show()
```
